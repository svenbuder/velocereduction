```python
# Detector Characterisation for CCD1, 2, and 3, for both 2Amp and 4Amp modes

from velocereduction import detector

gain_ccd_amp_texp = [
    [1,2,60],
    [1,4,60],
    [2,2,1],
    [2,2,10],
    [2,4,1],
    [2,4,10],
    [3,2,0.1],
    [3,4,0.1]
]

for ccd, amp, texp in gain_ccd_amp_texp:

    if amp == 2:
        date = 240219; file_begin="19feb"
    elif amp == 4:
        date = 260703; file_begin="03jul"

    if (ccd == 1) & (amp == 2):
        files = ['0030', '0031', '0032', '0033', '0034', '0035']
    elif (ccd == 2) & (amp == 2) & (texp == 1):
        files = ['0016', '0017', '0018', '0019', '0020', '0021']
    elif (ccd == 2) & (amp == 2) & (texp == 10):
        files = ['0023', '0024', '0025', '0026', '0027', '0028']
    elif (ccd == 3) & (amp == 2):
        files = ['0009', '0010', '0011', '0012', '0013', '0014']
    elif (ccd == 1) & (amp == 4):
        files = ['0035', '0036', '0037', '0038', '0039', '0040']
    elif (ccd == 2) & (amp == 4) & (texp == 1):
        files = ['0021', '0022', '0023', '0024', '0025', '0026']
    elif (ccd == 2) & (amp == 4) & (texp == 10):
        files = ['0028', '0029', '0030', '0031', '0032', '0033']
    elif (ccd == 3) & (amp == 4):
        files = ['0014', '0015', '0016', '0017', '0018', '0019']
    else:
        raise ValueError(f"CCD {ccd} and Amp {amp} combination not supported.")

    print('Characterising gain for CCD', ccd, 'and', amp, 'Amp with texp', texp, 's from files:', files, ' for night ', date)

    flat_files = [f"../../../observations/{date}/ccd_{ccd}/{file_begin}{ccd}{file}.fits" for file in files]

    gains, diagnostics = detector.characterise_detector_gain(
        flat_files,
        ccd=ccd,
        output_file=f"detector_gains_ccd{ccd}_{amp}amp_{texp}s.ecsv",
    )

    print(gains[['amplifier','gain_e_per_adu','gain_err_e_per_adu']])
    print(gains[['amplifier','read_noise_adu','read_noise_err_adu','overscan_rms_adu']])

    detector.plot_gain_characterisation(
        diagnostics,
        ccd=ccd,
        date=date,
        texp=texp,
        files=files,
        output_directory="./",
    )
```
